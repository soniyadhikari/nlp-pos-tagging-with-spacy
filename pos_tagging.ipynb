```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp('My name is Soniya Adhikari.')
```

```python
doc.text
```

```python
doc[0]
```

```python
doc[3].pos_ #coarse grain pos
```

```python
doc[0].tag_ # fine grain pos
```

```python
spacy.explain('PRP')
```

```python
for word in doc:
  print(word.text, ':', word.pos_, word.tag_, spacy.explain(word.tag_))
```

```python
doc2  =nlp('I left the school.')
for word in doc2:
  print(word.text, ':', word.pos_, word.tag_, spacy.explain(word.tag_))
```

```python
doc3  =nlp('to the left of the room')
for word in doc3:
  print(word.text, ':', word.pos_, word.tag_, spacy.explain(word.tag_))
```

```python
doc4  =nlp('I read books on history')
for word in doc4:
  print(word.text, ':', word.pos_, word.tag_, spacy.explain(word.tag_))
```

```python
doc6 = nlp('The quick brown fox jumped over the lazy dog.')
```

```python
from spacy import displacy
displacy.render(doc6, style='dep', jupyter=True)
```

```python
options = {'compact': True, 'distance': 80, 'color':'#fff', 'bg':'#00a65a'}
displacy.render(doc6, style='dep', jupyter=True, options=options)
```
